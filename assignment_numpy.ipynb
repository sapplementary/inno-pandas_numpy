**1.** Given a random array, change the sign of elements whose values ​​are between 3 and 8

```python
import numpy as np

x = np.random.randint(0, 12, 10)
mask = (x >= 3) & (x <= 8)
x[mask] *= -1

print(x)
```

**2.** Replace the maximum element of a random array with 0

```python
import numpy as np

x = np.random.randint(0, 12, 10)
print("before:   ", x)

i = x.argmax()
x[i] = 0

print("after:", x)
print("changed index:", i)
```

**3.** Construct the direct product of arrays (all combinations with each element). A two-dimensional array is provided as input.

```python
X = np.array([[1, 2], [3, 4]])

grids = np.meshgrid(*X, indexing='ij')
result = np.array(grids).reshape(len(X), -1).T

print(result)
```

**4.** Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.

```python
import numpy as np

A = np.random.randint(0, 5, (8, 3))
B = np.random.randint(0, 5, (2, 2))

eq = A[:, None, :, None] == B[None, :, None, :]
ok = eq.any(axis=2).all(axis=(1, 2))

result = A[ok]

print("A:\n", A)
print("B:\n", B)
print("mask:", ok)
print("result:\n", result)
```

**5.** Given a 10x3 matrix, find rows of unequal values ​​(for example, the row [2,2,3] remains, the row [3,3,3] is deleted)

```python
import numpy as np

A = np.random.randint(0, 3, (10, 3))

mask = A.min(axis=1) != A.max(axis=1)

result = A[mask]

print("A:\n", A)
print("mask:", mask)
print("result:\n", result)
```

**6.** Given a two-dimensional array, remove the rows that are repeated.

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
import numpy as np

def task1_pure(X):
    result = 1

    for i in range(min(len(X), len(X[0]))):
        if X[i][i] != 0:
            result *= X[i][i]
    return result

def task1_np(X):
    d = np.diag(X)
    return np.prod(d[d != 0])

X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
print(task1_pure(X.tolist()), task1_np(X))
```

* __Task 2__: Two vectors x and y are given. Check if they define the same multiset.  
  For example, for x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) answer is True.

```python
import numpy as np

def task2_pure(x, y):
    return sorted(x) == sorted(y)

def task2_np(x, y):

    return np.array_equal(np.sort(x), np.sort(y))

x, y = np.array([1, 2, 2, 4]), np.array([4, 2, 1, 2])
print(task2_pure(x.tolist(), y.tolist()), task2_np(x, y))
print(task2_np(x, np.array([4, 2, 1, 1])))
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero. 
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
def task3_pure(x):
    best = None
    for i in range(1, len(x)):
        if x[i - 1] == 0 and (best is None or x[i] > best):
            best = x[i]
    return best

def task3_np(x):

    return x[1:][x[:-1] == 0].max()

x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
print(task3_pure(x.tolist()), task3_np(x))
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
import numpy as np

def task4_pure(x):
    values, counts = [x[0]], [1]
    for v in x[1:]:
        if v == values[-1]:
            counts[-1] += 1
        else:
            values.append(v)
            counts.append(1)
    return values, counts

def task4_np(x):
    change = np.flatnonzero(np.diff(x) != 0) + 1
    starts = np.r_[0, change]
    lengths = np.diff(np.r_[starts, len(x)]) 
    return x[starts], lengths

x = np.array([2, 2, 2, 3, 3, 3, 5])
print(task4_pure(x.tolist()))
print(task4_np(x))
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
import math, time
from scipy.spatial.distance import cdist

def task5_pure(X, Y):

    return [[math.sqrt(sum((a - b) ** 2 for a, b in zip(p, q))) for q in Y] for p in X]

def task5_np(X, Y):
    return np.sqrt(((X[:, None, :] - Y[None, :, :]) ** 2).sum(axis=2))


X, Y = np.random.rand(5, 3), np.random.rand(4, 3)
assert np.allclose(task5_np(X, Y), cdist(X, Y))
assert np.allclose(task5_pure(X.tolist(), Y.tolist()), cdist(X, Y))

X, Y = np.random.rand(300, 10), np.random.rand(300, 10)
Xl, Yl = X.tolist(), Y.tolist()

for name, f in [("pure python", lambda: task5_pure(Xl, Yl)),
                ("numpy",       lambda: task5_np(X, Y)),
                ("cdist",       lambda: cdist(X, Y))]:
    t = time.perf_counter()
    f()
    print(f"{name:12s} {time.perf_counter() - t:.5f} c")
```

* #### __Task 6__: CrunchieMunchies __*__

You work in the marketing department of a food company called MyCrunch, which is developing a new type of delicious, healthy cereal called **CrunchieMunchies**.

You want to show consumers how healthy your cereal is compared to other leading brands, so you've compiled nutrition data from several different competitors.

Your task is to use Numpy calculations to analyze this data and prove that your **СrunchieMunchies** - the healthiest choice for consumers.


```python
import numpy as np
```

1. View the cereal.csv file. This file contains calorie counts for various brands of cereal. Download the data from the file and save it as calorie_stats.

```python
calorie_stats = np.loadtxt("./data/cereal.csv", delimiter=",")
calorie_stats
```

2. One serving of CrunchieMunchies contains 60 calories. How much higher is the average calorie count of your competitors?

Store the answer in the variable average_calories and print the variable in the terminal

```python
average_calories = calorie_stats.mean() - 60
print("average_calories:", average_calories)
```

3. Does the average calorie count accurately reflect the distribution of the data set? Let's sort the data and see.

Sort the data and store the result in the calorie_stats_sorted variable. Print the sorted information.

```python
calorie_stats_sorted = np.sort(calorie_stats)
print("sorted:", calorie_stats_sorted)
```

4. It looks like most values ​​are above the mean. Let's see if the median is the most accurate measure of the data set.

Calculate the median of the data set and save your answer in median_calories. Print the median so you can see how it compares to the mean.

```python
median_calories = np.median(calorie_stats)
print("median_calories:", median_calories)
```

5. While the median shows that at least half of our values ​​are over 100 calories, it would be more impressive to show that a significant portion of the competitors have higher calorie counts than CrunchieMunchies.

Calculate various percentiles and print them until you find the lowest percentile that exceeds 60 calories. Store this value in the nth_percentile variable.

```python
nth_percentile = None
for n in range(1, 101):
    value = np.percentile(calorie_stats, n)
    print(n, value)
    if value > 60:
        nth_percentile = n
        break
print("nth_percentile:", nth_percentile)
```

6. While the percentile shows us that most competitors have much higher calorie counts, it's an awkward concept to use in marketing materials.

Instead, let's calculate the percentage of cereals that contain more than 60 calories per serving. Save your answer in the more_calories variable and print it.

```python
more_calories = (calorie_stats > 60).mean() * 100
print("more_calories, %:", more_calories)
```

7. This is a really high percentage. This will be very useful as we promote CrunchieMunchies. But one question is: how big are the differences in the dataset? Can we generalize that most cereals contain around 100 calories or is the range even wider?

Calculate the magnitude of the deviation by finding the standard deviation. Save your answer in calorie_std and print it on the terminal. How can we incorporate this value into our analysis?

```python
'''The standard deviation (19.4) indicates how much the caloric content of individual cereal samples differs,
on average, from the mean (~107 kcal).A higher value signifies a greater spread in the data.
The interval defined as the mean ± 1 standard deviation (87.5–126.2) represents a "typical" range,
which encompasses the majority of the products. A value of 60 falls below this range
(approximately 2.4 standard deviations from the mean),meaning that 60 kcal is a notably atypically low caloric value for this dataset.'''

calorie_std = np.std(calorie_stats)
print("calorie_std:", calorie_std)
```

8. Please write a short paragraph outlining your findings and how you believe this data can be used to Mycrunch's advantage in marketing CrunchieMunchies.

```python
low = calorie_stats.mean() - calorie_std
high = calorie_stats.mean() + calorie_std
mean = calorie_stats.mean()

print(f"mean: {mean:.1f}")
print(f"60 is {(mean - 60) / calorie_std:.1f} std below the mean")  # ≈ 2.4
print(f"typical range: {low:.1f} - {high:.1f}")
print("60 below the range:", 60 < low)
```

The analysis shows that the average cereal contains about 107 calories per serving, and most products fall within a typical range of roughly 87 to 126 calories (one standard deviation around the mean). CrunchieMunchies, at 60 calories, sits clearly below this range, which makes it noticeably lighter than the typical competitor on the market. Mycrunch can use this in its marketing by positioning CrunchieMunchies as a low-calorie breakfast option for health-conscious customers, supporting the claim with the actual comparison to the market average. Because the difference is large relative to the spread of the data (about 2.4 standard deviations below the mean), this is not a marginal difference but a genuine competitive advantage that can be highlighted on packaging and in advertising.
